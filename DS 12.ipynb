# Part-1

## 1.

### We will apply Z-test in this scenario because the population standard deviations are known to us as well as the sample sizes for both rural and urban cases are large enough (>30)

## 2.

```python
import numpy as np
from scipy.stats import norm

urban_mean = 70
rural_mean = 74
urban_std = 11
rural_std = 9
n_urban = 85
n_rural = 90

SE = np.sqrt((urban_std**2 / n_urban) + (rural_std**2 / n_rural))

Z = (urban_mean - rural_mean) / SE
print(f"Z-statistic: {Z:.2f}")

alpha_2tailed_01 = norm.ppf(1-0.01/2)  
alpha_2tailed_05 = norm.ppf(1-0.05/2)
alpha_1tailed_01 = norm.ppf(1-0.01)
alpha_1tailed_05 = norm.ppf(1-0.05) 

if abs(Z) > alpha_2tailed_01:
    print("Two-tailed test at alpha = 0.01: Reject H0 (There's a significant difference between rural and urban satisfaction)")
else:
    print("Two-tailed test at alpha = 0.01: Fail to reject H0")

if abs(Z) > alpha_2tailed_05:
    print("Two-tailed test at alpha = 0.05: Reject H0 (There's a significant difference between rural and urban satisfaction)")
else:
    print("Two-tailed test at alpha = 0.05: Fail to reject H0")

if Z < -alpha_1tailed_01:
    print("One-tailed test at alpha = 0.01: Reject H0 (Urban customers are unsatisfied than Rural ones)")
else:
    print("One-tailed test at alpha = 0.01: Fail to reject H0")

if Z < -alpha_1tailed_05:
    print("One-tailed test at alpha = 0.05: Reject H0 (Urban customers are unsatisfied than Rural ones)")
else:
    print("One-tailed test at alpha = 0.05: Fail to reject H0")
```

## 1.

Conditions for a valid Chi-Square test:

Categorical Data: Both variables must be categorical for the test to apply.

Independence: Observations must be independent, with no influence between categories.

Expected Frequency: Each expected frequency should be at least 5 to ensure reliability.

Sample Size: A sufficiently large sample size is required for valid results.

## 2.

```python
import numpy as np
from scipy.stats import chi2_contingency, chi2

data = np.array([[55, 45], [75, 25]])

chi2_stat, p_val, dof, expected = chi2_contingency(data)

print(f"Chi-Square statistic: {chi2_stat:.2f}")
print(f"P-value: {p_val:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Expected frequencies:\n{expected}")

alpha_2tailed_01 = chi2.ppf(1 - 0.01 / 2, dof)
alpha_2tailed_05 = chi2.ppf(1 - 0.05 / 2, dof)

alpha_1tailed_01 = chi2.ppf(1 - 0.01, dof)
alpha_1tailed_05 = chi2.ppf(1 - 0.05, dof)

if chi2_stat > alpha_2tailed_01:
    print("Two-tailed test at alpha = 0.01: Reject H0 (There is a significant association between shopping format and satisfaction)")
else:
    print("Two-tailed test at alpha = 0.01: Fail to reject H0")

if chi2_stat > alpha_2tailed_05:
    print("Two-tailed test at alpha = 0.05: Reject H0 (There is a significant association between shopping format and satisfaction)")
else:
    print("Two-tailed test at alpha = 0.05: Fail to reject H0")

if chi2_stat > alpha_1tailed_01:
    print("One-tailed test at alpha = 0.01: Reject H0 (Offline format has higher satisfaction)")
else:
    print("One-tailed test at alpha = 0.01: Fail to reject H0")

if chi2_stat > alpha_1tailed_05:
    print("One-tailed test at alpha = 0.05: Reject H0 (Offline format has higher satisfaction)")
else:
    print("One-tailed test at alpha = 0.05: Fail to reject H0")
```

## 4.

Limitations of Chi-Square Test

Sample Size: Works best with large samples; small sample sizes can lead to unreliable results due to low expected frequencies.

Expected Frequency: Assumes each expected frequency is at least 5; violations may require an alternative test like Fisher's Exact Test.

No Direction: Only indicates association, not the direction or nature of the relationship, requiring additional tests to assess specifics.
