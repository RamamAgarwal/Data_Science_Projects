```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv('student_dataset-9B.csv')
```

```python
df
```

# Part-1

```python
contingency_table = pd.crosstab(df['Major'], df['Status'])
print(contingency_table)
```

```python
chi2_stat, p_value, dof, expected_freq = chi2_contingency(contingency_table)
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-Value: {p_value}")
print(f"Degrees of Freedom: {dof}")
print(f"Expected Frequencies:\n{expected_freq}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('There is no significant relationship between Major and Status of enrolment.')
elif p_value >= alpha:
    print('There is a significant relationship between Major and Status of enrolment.')
```

# Part-2

```python
df.groupby('Major')['Credits Earned'].mean()
```

```python
mean_credits = df.groupby('Major')['Credits Earned'].mean().reset_index()
print(mean_credits['Credits Earned'])
```

```python
overall_mean = df['Credits Earned'].mean()

SStotal = sum((df['Credits Earned'].values - overall_mean) ** 2)
print(f"SStotal: {SStotal}")
```

```python
SSG = sum(df.groupby('Major').size() * (mean_credits['Credits Earned'].values - overall_mean) ** 2)
print(f"SSG: {SSG}")
```

```python
SSE = sum((df['Credits Earned'] - df.groupby('Major')['Credits Earned'].transform('mean')) ** 2)
print(f"SSE: {SSE}")
```

## Interpretation

SStotal: This value indicates the total variability in the 'Credits Earned' across all observations. A higher value indicates more variability.

SSG (Between Groups): This value indicates the variability due to the differences between the group means. If SSG is high as compared to SStotal, it suggests that the means of the different Majors are significantly different.

SSE (Within Groups): This value indicates the variability within each group. A lower SSE compared to SStotal indicates that most of the variability is between the groups rather than within them.
